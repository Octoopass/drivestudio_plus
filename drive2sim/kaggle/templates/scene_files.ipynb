```python
# Tìm cảnh trong output notebook ingest đã gắn (theo manifest.yaml, không giả định đường gắn của
# Kaggle), rồi in file nhỏ của nó vào log (stderr, có trong `kaggle kernels logs`).
import base64
import json
import os
import pathlib
import sys

sys.path.insert(0, "/kaggle/tmp/repo")
from drive2sim.kaggle.scene_files import light_scene_files, write_log_lines

scene_id = "$scene_id"
found = sorted(pathlib.Path("/kaggle/input").glob(f"**/data/{scene_id}/manifest.yaml"))
assert len(found) == 1, f"tìm thấy {len(found)} cảnh {scene_id} dưới /kaggle/input: {found}"
files = light_scene_files(found[0].parent)
print(scene_id, len(files), "file")
light = {"scene_id": scene_id, "files": files}
write_log_lines("D2S_SCENE_FILES", light)  # dòng ngắn đánh số, stdout không chen vào được
```
